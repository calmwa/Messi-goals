```python

import pandas as pd
import matplotlib.pyplot as plt

url = "https://raw.githubusercontent.com/calmwa/Messi-goals/main/data/messi_goals.csv"
df = pd.read_csv(url)

print("Total goal records:", len(df))
print("Columns:", df.columns.tolist())
display(df.head())
```

```python

goals_per_season = df.groupby("Season").size().sort_index()

plt.figure(figsize=(12, 5))
plt.plot(goals_per_season.index, goals_per_season.values, marker="o")
plt.title("Lionel Messi Goals by Season")
plt.xlabel("Season")
plt.ylabel("Number of Goals")
plt.xticks(rotation=45)
plt.grid(True)
plt.tight_layout()
plt.show()
```

### Visualization 1: Messi's Goals by Season

**Research Question:** How did Lionel Messi's goal-scoring performance change throughout the seasons represented in the dataset?

**Methodology:** Using Python, Pandas, and Matplotlib, we grouped Messi's recorded goals by season and displayed the totals in a line graph.

**Purpose:** This visualization helps identify changes in Messi's goal-scoring output, including his highest-scoring seasons and how his performance varied over time.

**Limitations:** The analysis only covers goals included in our dataset. It does not represent his entire career.
